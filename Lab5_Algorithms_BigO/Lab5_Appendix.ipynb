```python
# [INITIALIZATION] 
LAST_NAME = "Atienza"
STUDENT_ID = "TUPM-26-1242"
FAVORITE_ARTIST = "IV OF SPADES"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))

# Calculate ARTIST_SUM (A=1 to Z=26)
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())

print(f"Seed: {SEED_DIGIT} | ID Sum: {ID_SUM} | Name Len: {NAME_LENGTH}")
print(f"Artist Len: {ARTIST_LENGTH} | Artist Sum: {ARTIST_SUM}")
```

```python
# [EXERCISE 1] EMERGENCY DISPATCH RECORD SEARCH
original_records = [100 + (i * ID_SUM) for i in range(6)]
rotation_point = SEED_DIGIT if 0 < SEED_DIGIT < len(original_records) else 2
rotated_records = original_records[-rotation_point:] + original_records[:-rotation_point]
target_incident = original_records[(ID_SUM + ARTIST_SUM) % len(original_records)]

def linear_search_count(data, target):
    comparisons = 0
    for i, val in enumerate(data):
        comparisons += 1
        if val == target: return i, comparisons
    return -1, comparisons

def rotated_binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        comparisons += 1
        if data[low] <= data[mid]:
            comparisons += 2
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            comparisons += 2
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

lin_idx, lin_comp = linear_search_count(rotated_records, target_incident)
bin_idx, bin_comp = rotated_binary_search(rotated_records, target_incident)

print("Original Dataset:", original_records)
print("Rotated Dataset:", rotated_records)
print(f"Target: {target_incident} | Linear Index: {lin_idx} (Comps: {lin_comp}) | Rotated Binary Index: {bin_idx} (Comps: {bin_comp})")
print("Complexity -> Time: O(log n) Binary, O(n) Linear | Space: O(1)")
```

```python
# [EXERCISE 2] UNIVERSITY ENROLLMENT RECORDS
dataset_A = [SEED_DIGIT * 100 + (i * ID_SUM) for i in range(5)]
dataset_B = [ARTIST_LENGTH * 100 + (i * NAME_LENGTH) for i in range(5)]

def merge_sorted_records(arr1, arr2):
    consolidated = []
    comparisons = 0
    i, j = 0, 0
    while i < len(arr1) and j < len(arr2):
        comparisons += 1
        if arr1[i] < arr2[j]:
            consolidated.append(arr1[i])
            i += 1
        else:
            consolidated.append(arr2[j])
            j += 1
    while i < len(arr1):
        consolidated.append(arr1[i])
        i += 1
    while j < len(arr2):
        consolidated.append(arr2[j])
        j += 1
    return consolidated, comparisons

consolidated_dataset, merge_comps = merge_sorted_records(dataset_A, dataset_B)
is_ordered = all(consolidated_dataset[i] <= consolidated_dataset[i+1] for i in range(len(consolidated_dataset)-1))

print("Dept A Dataset:", dataset_A)
print("Dept B Dataset:", dataset_B)
print("Consolidated Dataset:", consolidated_dataset)
print(f"Records Before: A={len(dataset_A)}, B={len(dataset_B)} | After: {len(consolidated_dataset)} | Comparisons: {merge_comps} | Order Verified: {is_ordered}")
print("Complexity -> Time: O(n + m) | Space: O(n + m)")
```

```python
# [EXERCISE 3] E-COMMERCE ORDER PROCESSING
import time
import random

def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

def measure(sort_function, data, trials=3):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

base_orders = [(ID_SUM * i + ARTIST_SUM) % 100 for i in range(8)]
sorted_input = sorted(base_orders)
reverse_input = sorted(base_orders, reverse=True)
random_input = base_orders.copy()
random.seed(SEED_DIGIT)
random.shuffle(random_input)

conditions = {
    "Sorted": sorted_input,
    "Reverse": reverse_input,
    "Random": random_input
}

algorithms = {
    "Bubble": bubble_sort,
    "Selection": selection_sort,
    "Insertion": insertion_sort
}

print("Generated Orders:", base_orders)

for cond, dataset in conditions.items():
    print(f"\n--- {cond.upper()} INPUT: {dataset} ---")
    for name, func in algorithms.items():
        res, t = measure(func, dataset, trials=3)
        verified = "Pass" if all(res[0][i] <= res[0][i+1] for i in range(len(res[0])-1)) else "Fail"
        print(f"{name:<10} | Comps: {res[1]:<3} | Moves: {res[2]:<3} | Time: {t:.6f}s | Verify: {verified}")
```
